```python
import pandas as pd
```

```python
df = pd.read_csv(r"synthetic_online_retail_data.csv")
df.head(10)
```

```python
df.isna().sum()
```

```python
df['review_score'].fillna(df['review_score'].mean(), inplace=True)
```

```python
df['order_id'] = range(len(df))
```

customers_df → customer_id, gender, age, city 

products_df → product_id, product_name, category_name, price 

orders_df → order_id, customer_id, product_id, order_date, quantity, payment_method, review_score

```python
customers_df = df[['customer_id', 'gender', 'age', 'city']].drop_duplicates().reset_index(drop=True)
products_df = df[['product_id', 'product_name', 'category_name', 'price']].drop_duplicates().reset_index(drop=True)
orders_df = df[['order_id', 'customer_id', 'product_id', 'order_date', 'quantity', 'payment_method', 'review_score']].copy()
```

Remove duplicates in customer_df and products df

```python
# remove duplicates by id and reset index
products_before = len(products_df)
products_df.drop_duplicates(subset=['product_name'], keep='first', inplace=True)
products_df.reset_index(drop=True, inplace=True)

customers_before = len(customers_df)
customers_df.drop_duplicates(subset=['customer_id'], keep='first', inplace=True)
customers_df.reset_index(drop=True, inplace=True)

print(f"products: {products_before} -> {len(products_df)}; customers: {customers_before} -> {len(customers_df)}")
```

```python
unique_ids = df['product_id'].dropna().nunique()
unique_names = df['product_name'].dropna().nunique()
print(f"unique product_id: {unique_ids}, unique product_name: {unique_names}")

products_df = (
    df[['product_id', 'product_name', 'category_name', 'price']]
      .drop_duplicates(subset=['product_id'])
      .reset_index(drop=True)
)

print(f"products_df length after dedup: {len(products_df)}")
products_df.head(10)
```

```python
import sqlite3
import pandas as pd

conn = sqlite3.connect("../cleaned_data.db")

customers_df.to_sql("customers", conn, if_exists="replace", index=False)
orders_df.to_sql("orders", conn, if_exists="replace", index=False)
products_df.to_sql("products", conn, if_exists="replace", index=False)

tables = pd.read_sql(
    "SELECT name FROM sqlite_master WHERE type='table';",
    conn
)

print(tables)

print(pd.read_sql("SELECT COUNT(*) AS total_customers FROM customers", conn))
print(pd.read_sql("SELECT COUNT(*) AS total_orders FROM orders", conn))
print(pd.read_sql("SELECT COUNT(*) AS total_products FROM products", conn))

conn.commit()
conn.close()
```

```python
conn = sqlite3.connect("../cleaned_data.db")

query = """
SELECT SUM(o.quantity) AS total_smartphones_sold
FROM orders AS o
JOIN products AS p
    ON o.product_id = p.product_id
WHERE lower(p.category_name) LIKE '%smartphone%'
     OR lower(p.product_name) LIKE '%smartphone%'
"""

result = pd.read_sql_query(query, conn)
print(result)

conn.close()
```
